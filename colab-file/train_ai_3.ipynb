````python
from ultralytics import YOLO

# Load model (bạn có thể đổi yolov8n -> s/m/l)
model = YOLO("yolov8n.pt")

# Train
model.train(
    data="/content/fruit_dataset/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16
)
````
